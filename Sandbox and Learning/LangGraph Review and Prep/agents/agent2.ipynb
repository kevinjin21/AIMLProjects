# Multiple Inputs Graph
Handle multiple inputs and perform operations on list data.

```python
from typing import TypedDict, List
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    values: List[int]
    name: str
    result: str
```

```python
# Build our node
def process_values(state: AgentState) -> AgentState:
    """Handles and processes multiple inputs"""
    state['result'] = f"Hi there {state['name']}! Your sum is {sum(state['values'])}"
    return state # don't forget to pass on the updated state
```

```python
# Create graph
graph = StateGraph(AgentState)

graph.add_node('processor', process_values)
graph.set_entry_point('processor')  # Note: Use the node name, not the function name
graph.set_finish_point('processor')

app = graph.compile()   # Make sure you store the *compiled* graph in a variable
```

```python
result = app.invoke({"values": [1,2,3], "name": "Kevin"})
print(result)
result['result']
```

# Exercise 2
Create a graph that passes a list of ints, a name, and an operation. If operation is "+", add the elements, if it's a "\*", mult the elements. <br>
Input: {"name": "Jack Sparrow", "values": [1,2,3,4], "operation": "\*"} <br>
Output: "Hi Jack Sparrow, your answer is: 24"

```python
import math

# init AgentState
class AgentState(TypedDict):
    name: str
    values: List[int]
    operation: str
    result: str

# init graph
exercise = StateGraph(AgentState)

# create node and set edges
def calculate(state: AgentState) -> AgentState:
    """Return operation based on user input"""
    if state['operation'] not in ['+', '*']:
        raise ValueError("Invalid operation chosen.")
    if state['operation'] == '+':
        state['result'] = f"Hi {state['name']}, your answer is: {sum(state['values'])}"
    elif state['operation'] == '*':
        state['result'] = f"Hi {state['name']}, your answer is: {math.prod(state['values'])}"
    return state

exercise.add_node('calculator', calculate)
exercise.set_entry_point('calculator')
exercise.set_finish_point('calculator')

app = exercise.compile()
```

```python
result = app.invoke({"name": "Jack Sparrow", "values": [1,2,3,4], "operation": "*"})
result['result']
```
