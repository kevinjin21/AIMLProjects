## 1. Implement k-means clustering: Write a function that takes a dataset and a number of clusters k and returns the cluster assignments for each data point.

#### Clarifying questions
- Inputs: Number of features, shape of input data, number of samples
- Hyperparams: Number of iterations, threshold to stop training
- Choose centroid randomly first ok? If not, what is preferred (i.e. some average)
- How to handle empty clusters? Re-init to random point ok? (related to previous question) Or reduce k?
- Is Euclidean distance ok for calculating similarity?
- Datatypes?

```python
"""Roadmap:
- Initialize k centroids randomly
- Calculate euclidean distances and assign clusters
- Calculate average of each cluster
- Repeat 2 and 3 for max_iterations or until threshold is reached
- Return labels and optionally centroids
"""
```

```python
import numpy as np

def kmeans(X, k, max_iter=100, tol=1e-4, random_state=42):
    """
    Calculate kmeans using euclidean distance
    X: (n, d) dataset array    # n data points(samples), d features
    k: number of clusters
    Returns: (n,) column of labels, centroids (k, d). Represents each final point chosen as a centroid (row in df)
    """
    # 1. Choose random centroid
    rng = np.random.default_rng(random_state)
    n = X.shape[0]  # number of rows
    idx = rng.choice(n, size=k, replace=False)  # k random indices in n rows
    C = X[idx].astype(float)    # make sure values are floats, optional

    # 4. Repeat steps 2 and 3 for max_iter
    for _ in range(max_iter):
        # 2. Assign clusters
        # (n, 1, d) - (1, k, d) -> (n, k) add dimensions and calc norm
        dists = np.linalg.norm(X[:, None, :] - C[None, :, :], axis=2)
        labels = dists.argmin(axis=1)   # assign clusters (for each point, find the nearest centroid)

        # 3. Find average and re-assign centroids (update)
        C_new = C.copy()
        for idx in range(k):    # for each cluster
            pts = X[labels == idx]  # find points in the cluster
            if len(pts) == 0:
                # re-init empty cluster to a random point
                C_new[idx] = X[rng.integers(0, n)]
            else:
                # re-assign to mean centroid
                C_new[idx] = pts.mean(axis=0)

        # 4b. Threshold
        shift = np.linalg.norm(C_new - C)   # how far new centroid is from old
        C = C_new   # Assign new centroid
        if shift < tol:
            break   # stop when shift is smaller than tolerance

    return labels, C    # point labels, centroids
```

## 2. Code gradient descent: Implement the gradient descent optimization algorithm from scratch to find the minimum of a simple function, often with only NumPy and SciPy allowed.

```python
"""
Clarifying questions:
- Fixed learning rate lr schedule
- Gradient given?
- Optimize for specific scalar function f, or general grad descent 
- Stopping criteria (max_iter or threshold based on gradient norm)
- x0 given or randomly chosen or array of zeroes
"""
```

```python
"""
Steps:
1. Start from x0 (if given)
2. Iteratively x <- x - lr * grad(x)
3. Stop when threshold is met or max_iter is reached
4. Return x, history
"""
```

```python
import numpy as np

def gradient_descent(f, grad, x0=None, lr=0.1, max_iter=100, tol=1e-6):
    """
    f: callable(x) -> scalar
    grad: callable(x) -> grad vector
    x0: starting point, if not given then random. Will need input dimension d as a result. Can then make a random array d or array of zeros d.
    """
    # 1. Start from x0
    x = np.array(x0, dtype=float)
    history = [] # init history (optional, but good for diagnosis

    # 2. Iterate
    for _ in range(max_iter):
        g = grad(x)
        history.append((x.copy(), f(x)))    # Store history
        if np.linalg.norm(g) < tol:
            break   # Threshold met
        x -= lr * g # step down gradient
    return x, history

# Example f and grad
f = lambda x: np.sum((x-3.0)**2) # (x-3)^T(x-e)
grad = lambda x: 2*(x-3.0)
```
