# Terms / Concepts Glossary
Reference: https://youtu.be/jGg_1h0qzaM?si=sCsefZRPtRVzlUjV

__State__ \
Shared data structure that holds current info or context for application. Similar to app's memory; keeps track of variables and data that nodes can reference.

__Node__ \
Individual functions or operations that perform tasks within the graph. Receives inputs from state, processes, and produces output.

__Graph__ \
Overall structure/map of workflow *visual representation/roadmap*.

__Edges__ \
Connections between nodes. Tell which node should come next (usually some kind of logic). \
Conditional Edge: Uses specific logic to decide the next node.

__START__ \
Virtual entry point, marks where the workflow begins. Doesn't perform additional function.

__Tools__ \
Specialized functions or utils that nodes can use to perform specific tasks i.e. API call, calculation, etc. 

__ToolNode__ \ 
Special kind of node whose main job is to run a tool.

__StateGraph__ \
Class in LangGraph used to manage nodes, edges, and state. Think blueprint of a building.

__Runnable__ \
Executable component that performs a task in an AI workflow. AI version of function.

__Messages__ 
- Human Message
- System Message
- AI Message
- Function Message
- Tool Message
